```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class BatsmenState(TypedDict):

    runs: int
    balls: int 
    fours: int
    sixes: int 

    sr: float
    bpb: float
    b_percentage : float
    summary: str
```

```python
def calculate_sr(state: BatsmenState) ->BatsmenState:
    sr= (state['runs']/state['balls'])*100

    return {'sr': sr}
```

```python
def calculate_bpb( state: BatsmenState ) -> BatsmenState:
    bpb = state['balls']/(state['fours'] + state['sixes'])
    return {'bpb' : bpb}
    
```

```python
def calculate_b_percentage(state:BatsmenState) -> BatsmenState:
    b_percent = (((state['fours'] * 4) + (state['sixes'] * 6))/ state['runs'])*100
    return {'b_percentage': b_percent}
    
```

```python
def summary (state: BatsmenState) -> BatsmenState:
    summary = f"""
    Strike Rate - {state['sr']} \n
    Balls per Boundary - {state['bpb']}\n
    Boundary percentage - {state['b_percentage']}
    """

    return {'summary': summary}
```

```python
graph = StateGraph(BatsmenState)

#nodes
graph.add_node('calculate_sr', calculate_sr)
graph.add_node('calculate_bpb', calculate_bpb)
graph.add_node('calculate_b_percentage', calculate_b_percentage)
graph.add_node('summary', summary)

#edges
graph.add_edge(START,'calculate_sr')
graph.add_edge(START,'calculate_bpb')
graph.add_edge(START,'calculate_b_percentage')

graph.add_edge('calculate_sr','summary')
graph.add_edge('calculate_bpb','summary')
graph.add_edge('calculate_b_percentage','summary')

graph.add_edge('summary', END)

workflow = graph.compile()
```

```python
initial_state = {
    'runs' : 100,
    'balls': 50,
    'fours': 6,
    'sixes':4
}

final_state = workflow.invoke(initial_state)
print(final_state)
```
